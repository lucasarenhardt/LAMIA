```python
import spacy
```

```python
nlp = spacy.load("en_core_web_md")
```

```python
with open("wiki_us.txt", 'r') as f:
    text = f.read()
```

```python
#imprimindo uma sentença

doc = nlp(text)
sentence1 = list(doc.sents)[0]
print(sentence1)
```

```python
import numpy as np


#procurando as 10 palavras que mais se relacionam com country
your_word = "country"

ms = nlp.vocab.vectors.most_similar(np.asarray([nlp.vocab.vectors[nlp.vocab.strings[your_word]]]), n=10)
words = [nlp.vocab.strings[w] for w in ms[0][0]]
distances = ms[2]
print(words)
```

```python
doc1 = nlp("I like salty fries and hamburgers.")
doc2 = nlp("Fast food tastes very good.")
```

```python
#verificando a similaridade entre as 2 frases

print(doc1, "<->", doc2, doc1.similarity(doc2))
```
