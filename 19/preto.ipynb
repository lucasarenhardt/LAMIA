```python
import cv2 as cv
import numpy as np
```

```python
#criando um objeto de captura

cap = cv.VideoCapture(0)
```

```python
while True:

    #armazenando os frames
    _, frame = cap.read()
    hsv_frame = cv.cvtColor(frame, cv.COLOR_BGR2HSV)

    #criando a máscara com o intervalo desejado
    low_black = np.array([0, 0, 0])
    high_black = np.array([255, 255, 30])
    mask = cv.inRange(hsv_frame, low_black, high_black)

    #mostrando a imagem
    cv.imshow("Frame", frame)
    cv.imshow("Black Mask", mask)

    #quebrando o loop
    key = cv.waitKey(1)
    if key == 27:
        cv.destroyAllWindows()
        break
        
```
