```python
#a função torch.ones cria um tensor com valores unitários

a = torch.ones(2,1,10)
a
```

```python
#a função squeeze exclui as dimensões de tamanho 1

print(f'Shape antes de utilizar squeeze: {a.shape}')
squeezed = a.squeeze(1)
print(f'Shape depois de utilizar squeeze: {squeezed.shape}')
```

```python
a = torch.ones(2,3)
a
```

```python
#a função unsqueeze adiciona dimensões de tamanho 1

print(f'Shape antes de utilizar squeeze: {a.shape}')
squeezed = a.unsqueeze(2)
print(f'Shape depois de utilizar squeeze: {squeezed.shape}')
```

```python
#também é possível adicionar dimensões de tamanho 1 utilizando None

print(a[None].shape)
print(a[:,None].shape)
print(a[:,:,None].shape)
```

```python
#multiplicação de matrizes

x = torch.tensor([[1,2,3,4],[5,6,7,8]])
y = torch.tensor([[1,2,3],
                  [2,3,4],
                  [4,5,6],
                  [7,8,9]])

torch.matmul(x,y)
```

```python
#concatenação

x = torch.randn(1,4,5)
z = torch.cat([x,x], axis=1)
print('Eixo 1 concatenado: ', x.shape, z.shape)
```

```python
#permutação

x = torch.randn(3,20,10)
z = x.permute(2,0,1)
print('Dimensões trocadas:', x.shape, z.shape)
```
