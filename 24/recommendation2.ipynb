```python
import pandas as pd

#carregando e unindo os dados em um dataset
r_cols = ['user_id', 'movie_id', 'rating']
ratings = pd.read_csv('data/u.data', sep='\t', names=r_cols, usecols=range(3))

m_cols = ['movie_id', 'title']
movies = pd.read_csv('data/u.item', sep='|', names=m_cols, usecols=range(2), encoding='ISO-8859-1')

ratings = pd.merge(movies, ratings)

ratings.head()
```

```python
#modificando o formato do dataset

movieRatings = ratings.pivot_table(index=['user_id'], columns=['title'], values=['rating'])
movieRatings.head()
```

```python
movieRatings.columns = movieRatings.columns.get_level_values(1)
starWarsRatings = movieRatings['Star Wars (1977)'] # avaliação do star wars 
starWarsRatings
```

```python
#construindo a correlação entre star wars e os outros filmes, removendo os valores nulos
#e transformando isso num dataset

similarMovies = movieRatings.corrwith(starWarsRatings)
similarMovies = similarMovies.dropna()
df = pd.DataFrame(similarMovies)
df.head(10)
```

```python
#ordenando os filmes por similaridade
similarMovies.sort_values(ascending=False)
```

```python
import numpy as np

#criando um novo dataset com o número de pessoas que avaliaram o filme e a média das avaliações
movieStats = ratings.groupby('title').agg({'rating': ['size', 'mean']})
movieStats.head()
```

```python
#filtrando apenas os filmes que foram avaliados por mais de 100 pessoas
popularMovies = movieStats['rating']['size'] >= 100
movieStats[popularMovies].sort_values([('rating', 'mean')], ascending=False)[:15]
```

```python
movieStats[popularMovies]['rating']
```

```python
#unindo o dataframe com a similaridade
df = movieStats[popularMovies]['rating'].join(pd.DataFrame(similarMovies, columns=['similarity']))
df.head()
```

```python
df.sort_values(['similarity'], ascending=False)[:15]
```

```python
import pandas as pd

#carregando os dados pro dataset
r_cols = ['user_id', 'movie_id', 'rating']
ratings = pd.read_csv('data/u.data', sep='\t', names=r_cols, usecols=range(3))

m_cols = ['movie_id', 'title']
movies = pd.read_csv('data/u.item', sep='|', names=m_cols, usecols=range(2), encoding='ISO-8859-1')

ratings = pd.merge(movies, ratings)
ratings.head()
```

```python
#modificando o formato do dataset
userRatings = ratings.pivot_table(index=['user_id'], columns=['title'], values='rating')
userRatings.head()
```

```python
#matriz de correlação entre as avaliações dos filmes
corrMatrix = userRatings.corr()
corrMatrix.head()
```

```python
#filtrando apenas os filmes que forma avaliados por mais de 100 pessoas
corrMatrix = userRatings.corr(method='pearson', min_periods=100)
corrMatrix.head()
```

```python
#removendo os valores nulos
myRatings = userRatings.loc[0].dropna()
myRatings
```

```python
simCandidates = pd.Series() 

#adicionar os filmes similares
for i in range(0, len(myRatings.index)):
    print('Adding sims for ' + str(myRatings.index[i]) + '...')
    sims = corrMatrix[myRatings.index[i]].dropna()
    sims = sims.map(lambda x: x * myRatings[i])
    simCandidates = simCandidates._append(sims)

print('Sorting...')

#ordenando e mostrando os filmes
simCandidates.sort_values(inplace=True, ascending=False)
print(simCandidates.head(10))
```

```python
#utilizando o gorupby para somar a nota dos filmes que aparecem mais de uma vez
simCandidates = simCandidates.groupby(simCandidates.index).sum() 
simCandidates.sort_values(inplace=True, ascending=False)
simCandidates.head(10)
```

```python
filteredSims = simCandidates.drop(myRatings.index)
filteredSims
```
