```python
import tensorflow as tf
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
```

```python
#criando algumas frases

sentences = [
    "Eu gosto de cachorro e gato.",
    "Eu quero um carro e um videogame.",
    "Eu odeio o Twitter."
]
```

```python
#transformando as palavras em números inteiros

MAX_VOCAB_SIZE = 20000
tokenizer = Tokenizer(num_words=MAX_VOCAB_SIZE)
tokenizer.fit_on_texts(sentences)
sequences = tokenizer.texts_to_sequences(sentences)
```

```python
print(sequences)
```

```python
#dicionário com as palavras

tokenizer.word_index
```

```python
#realizando o padding

data = pad_sequences(sequences)
print(data)
```

```python
#colocando o padding no final

data = pad_sequences(sequences, padding='post')
print(data)
```

```python
#truncando 

data = pad_sequences(sequences, maxlen=3)
print(data)
```

```python
data = pad_sequences(sequences, maxlen=3, truncating='post')
print(data)
```
