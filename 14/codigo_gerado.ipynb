```python
import keras
import tensorflow as tf
```

```python
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, BatchNormalization, Dense, Dropout, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras import utils as np_utils
import matplotlib.pyplot as plt
```

```python
#carregando o dataset
(X_treinamento, y_treinamento), (X_teste, y_teste) = cifar10.load_data()
```

```python
#redimensionando e normalizando os dados
X_treinamento = X_treinamento.reshape(X_treinamento.shape[0], 32, 32, 3)
X_teste = X_teste.reshape(X_teste.shape[0], 32, 32, 3)
X_treinamento = X_treinamento.astype('float32')
X_teste = X_teste.astype('float32')
X_treinamento /= 255
X_teste /= 255
```

```python
#transformando para o formato onehot encoded
y_treinamento = np_utils.to_categorical(y_treinamento, 10)
y_teste = np_utils.to_categorical(y_teste, 10)
```

```python
#criando as camadas da rede neural
rede_neural = Sequential()

#camada de entrada
rede_neural.add(InputLayer(shape=(32, 32, 3)))

#camada de convolução
rede_neural.add(Conv2D(filters=32, kernel_size=(2, 2), activation='relu'))
#camada de normalização
rede_neural.add(BatchNormalization())
#camada de pooling
rede_neural.add(MaxPooling2D(pool_size=(2, 2)))
#camada de convolução
rede_neural.add(Conv2D(filters=32, kernel_size=(2, 2), activation='relu'))
#camada de normalização
rede_neural.add(BatchNormalization())
#camada de pooling
rede_neural.add(MaxPooling2D(pool_size=(2, 2)))

#camada de flattening
rede_neural.add(Flatten())

#camada oculta
rede_neural.add(Dense(units=256, activation='relu'))
#camada de dropout
rede_neural.add(Dropout(0.3))
#camada oculta
rede_neural.add(Dense(units=32, activation='relu'))
#camada de dropout
rede_neural.add(Dropout(0.3))
#camada de saída
rede_neural.add(Dense(units=10, activation='softmax'))
```

```python
rede_neural.summary()
```

```python
#compilando a rede neural
rede_neural.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])
```

```python
#realizando o treinamento
rede_neural.fit(X_treinamento, y_treinamento, batch_size=128, epochs=10, validation_data=(X_teste, y_teste))
```

```python
resultado = rede_neural.evaluate(X_teste, y_teste)
resultado
```
