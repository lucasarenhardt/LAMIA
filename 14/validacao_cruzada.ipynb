```python
from tensorflow.keras.datasets import mnist
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import InputLayer, Dense, Flatten, Conv2D, MaxPooling2D
from tensorflow.keras import utils as np_utils
import numpy as np
from sklearn.model_selection import StratifiedKFold
```

```python
(X, y), (X_teste, y_teste) = mnist.load_data()
X = X.reshape(X.shape[0], 28, 28, 1)
X = X.astype('float32')
X /= 255
y = np_utils.to_categorical(y, 10)
```

```python
seed = 5
np.random.seed(seed)
```

```python
kfold = StratifiedKFold(n_splits=5, shuffle=True, random_state=seed)
```

```python
resultados = []
for indice_treinamento, indice_teste in kfold.split(X, np.zeros(shape=(y.shape[0], 1))):
  print("Indices treinamento: ", indice_treinamento, "Indices teste: ", indice_teste)
  #criando as camadas da rede neural
  classificador = Sequential()

  #camada de entrada
  classificador.add(InputLayer(shape=(28, 28, 1)))

  #camada de convolução
  classificador.add(Conv2D(filters=32, kernel_size=(3, 3), activation='relu'))
  #camada de pooling
  classificador.add(MaxPooling2D(pool_size=(2, 2)))

  #camada de flattening
  classificador.add(Flatten())

  #camada oculta
  classificador.add(Dense(units=128, activation='relu'))
  #camada de saída
  classificador.add(Dense(units=10, activation='softmax'))

  #compilando a rede neural
  classificador.compile(loss='categorical_crossentropy', optimizer='adam', metrics=['accuracy'])

  #realizando o treinamento
  classificador.fit(X[indice_treinamento], y[indice_treinamento], epochs=5, batch_size=128)
  #calculando a precisão e adicionando à lista
  precisao = classificador.evaluate(X[indice_teste], y[indice_teste])
  resultados.append(precisao[1])
```
